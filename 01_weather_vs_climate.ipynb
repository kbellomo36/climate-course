# Weather vs Climate

### Learning objectives

By the end of this notebook, you should be able to:

- distinguish **weather** from **climate**;
- describe a time series of daily weather observations;
- calculate a simple climatology;
- identify variability around the climatological mean;
- understand why a single observation is not enough to characterize climate.

We will use a **synthetic daily temperature dataset** first. This keeps the notebook fast and reproducible; later lectures will use real observational datasets.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (11, 5)
plt.rcParams["font.size"] = 12

rng = np.random.default_rng(42)
```

## 1. Weather: one realization of the atmosphere

Let's create two years of daily temperature observations for an imaginary location.

The temperature contains:

1. a seasonal cycle;
2. day-to-day weather variability;
3. a small long-term warming signal.

The individual daily values represent **weather**.

```python
dates = pd.date_range("2000-01-01", "2001-12-31", freq="D")
day_of_year = dates.dayofyear.values

seasonal_cycle = 10 + 10 * np.sin(2 * np.pi * (day_of_year - 80) / 365.25)
weather_noise = rng.normal(0, 3, len(dates))
warming = np.linspace(0, 0.6, len(dates))

temperature = seasonal_cycle + weather_noise + warming

df = pd.DataFrame({"temperature": temperature}, index=dates)
df.head()
```

```python
ax = df["temperature"].plot(lw=0.8)
ax.set_ylabel("Temperature (°C)")
ax.set_xlabel("")
ax.set_title("Daily temperature: weather variability")
plt.show()
```

### Think about it

Look at the time series above.

- Can you identify individual weather events?
- Can you immediately tell whether the climate has changed?
- What features are predictable on long time scales?
- What features are mostly unpredictable from one day to the next?

The **weather** is the sequence of individual realizations.  
The **climate** emerges when we describe the statistical properties of many such observations.

## 2. From weather to climate

One way to describe climate is to calculate averages over a longer period.

First, let's compare the distribution of daily temperatures with its mean.

```python
mean_temperature = df["temperature"].mean()
std_temperature = df["temperature"].std()

print(f"Mean temperature: {mean_temperature:.2f} °C")
print(f"Standard deviation: {std_temperature:.2f} °C")

ax = df["temperature"].hist(bins=30)
ax.set_xlabel("Temperature (°C)")
ax.set_ylabel("Number of days")
ax.set_title("Distribution of daily temperatures")
plt.show()
```

The mean alone does not describe the climate completely.

Two locations could have the same mean temperature but very different variability.

This is why climate statistics include quantities such as:

- mean;
- variance;
- standard deviation;
- percentiles;
- extremes;
- trends.

## 3. Monthly climatology

A useful first climate statistic is the **monthly climatology**.

For each calendar month, we calculate the average temperature across all available years.

```python
monthly_climatology = df["temperature"].groupby(df.index.month).mean()

month_names = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

monthly_climatology.index = month_names
monthly_climatology
```

```python
ax = monthly_climatology.plot(marker="o")
ax.set_ylabel("Temperature (°C)")
ax.set_xlabel("Month")
ax.set_title("Monthly climatology")
plt.xticks(rotation=45)
plt.show()
```

### What have we done?

We started with **daily weather** and reduced it to a much simpler description:

> the typical temperature for each month.

That 12-month cycle is a simple **climatology**.

In real climate analysis, the reference period is important. We will later use standardized climate reference periods and real gridded observational datasets.

## 4. Weather versus climate: the key idea

Consider a very cold day in an otherwise warming climate.

The cold day is a **weather event**.

Whether the probability of such cold days is changing is a **climate question**.

Likewise, one exceptionally warm day does not by itself prove climate change. To identify climate change, we look for **persistent changes in the statistics of many observations**.
